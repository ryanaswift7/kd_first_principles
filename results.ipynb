# Narrow Layer Filtering Results

```python
up_offline_250 = pd.read_parquet(BASE_DIR + "narrow_layer_filtering/scores/results_df_20251025_061233.parquet")
rest_of_up_except_1000 = pd.read_parquet(BASE_DIR + "narrow_layer_filtering/scores/results_df_20251030_053339.parquet")
up_1000 = pd.read_parquet(BASE_DIR + "narrow_layer_filtering/scores/results_df_20251031_194315.parquet")
up_random = pd.read_parquet(BASE_DIR + "narrow_layer_filtering/scores/results_df_20251102_053240.parquet")
down_random = pd.read_parquet(BASE_DIR + "narrow_layer_filtering/scores/results_df_20251104_053028.parquet")
off_and_on = pd.read_parquet(BASE_DIR + "narrow_layer_filtering/scores/results_df_20251107_183137.parquet")

final_df = pd.concat([up_offline_250, rest_of_up_except_1000, up_1000], ignore_index=True)
filter_borked_random = ~final_df['EXPERIMENT'].str.contains("randomized")
final_df = final_df[filter_borked_random]
final_df = pd.concat([final_df, up_random, down_random, off_and_on]).sort_values(by='SCORE', ascending=False, ignore_index=True)
filter_base = ~final_df['EXPERIMENT'].str.contains("base")
final_df = final_df[filter_base]

final_df = final_df[["EXPERIMENT", "SCORE"]]

# NOTE: distilling all layers is at least as good as any of the other subsets
print(final_df.sort_values(by='EXPERIMENT').to_string(index=False))
```

## All Layers

# Offline Projector Search

## NOTE: UP-ALIGN_1.0 had many trial failures. May be possible to get (a likely very small amount of) extra performance out of it

```python
down_offline_align_000 = pd.read_parquet(BASE_DIR + "off_proj_search/scores/students/narrow-down_offline_align_0.0-balanced-jsd_0.5-filter_1.000/hellaswag/results_df_20251109_033234.parquet")
down_offline_align_050 = pd.read_parquet(BASE_DIR + "off_proj_search/scores/students/narrow-down_offline_align_0.5-balanced-jsd_0.5-filter_1.000/hellaswag/results_df_20251110_152543.parquet")
down_offline_align_025 = pd.read_parquet(BASE_DIR + "off_proj_search/scores/students/narrow-down_offline_align_0.25-balanced-jsd_0.5-filter_1.000/hellaswag/results_df_20251110_094340.parquet")

rest_of_down = pd.read_parquet(BASE_DIR + "off_proj_search/scores/results_df_20251114_153553.parquet")
up = pd.read_parquet(BASE_DIR + "off_proj_search/scores/results_df_20251114_193217.parquet")


off_proj_search_df = pd.concat([down_offline_align_000, down_offline_align_050, down_offline_align_025, rest_of_down, up], ignore_index=True)

print(off_proj_search_df[["EXPERIMENT", 'SCORE']].sort_values(by='EXPERIMENT').to_string(index=False))
```

### Another Note
- the best performer here was up-align_0.5 (0.371540)
- this is nearly a full half-percent better than the other up-align_0.5 trained for the layer filtering (0.365465 for all layers)
- while this is not an objectively large difference, narrow_ft sits in between these 2 values (0.370444)
- despite my attempts to make the training as reproducible as possible, this is evidence that there can be at least 0.005 difference between training runs (even with identical projectors)
